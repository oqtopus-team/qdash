# Investigate Results

Configure a `local` profile using the [Python reference](../../../docs/reference/clients/python.md). Install `pandas`, `pyarrow`, and `perspective-python[jupyter]` in the notebook environment. This notebook only reads data.

```python
from perspective.widget import PerspectiveWidget

from qdash.sdk import QDash

qdash = QDash.from_profile("local")
CHIP_ID = "YOUR_CHIP_ID"
```

## List Executions

```python
executions = qdash.executions.list(chip_id=CHIP_ID, limit=100)
print("Loaded:", len(executions), "Total:", executions.total)
PerspectiveWidget(executions.to_dataframe())
```

## Find Task Results

API filters search history. Sorting, filtering, and grouping in the widget apply only to the loaded page.

```python
results = qdash.task_results.list(chip_id=CHIP_ID, status="failed", limit=100)
print("Loaded:", len(results), "Total:", results.total)
PerspectiveWidget(results.to_dataframe())
```

## Next Page

`next_page()` preserves the API filters and returns `None` at the end. New records may shift offset-based pages.

```python
next_results = results.next_page()
next_results
```

## Inspect a Result

Copy a task ID from the table to inspect parameters and figures.

```python
task = qdash.task_result("YOUR_TASK_ID")
print(task.message)
print(task.input_parameters)
print(task.output_parameters)
for figure in task.figures:
    figure.show()
```

## Close Connection

```python
qdash.close()
```
