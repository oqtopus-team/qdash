# Investigate Results

Configure a `local` profile using the [Python reference](../../../docs/reference/clients/python.md). Install `pandas` in the notebook environment. This notebook only reads data.

```python
from IPython.display import display

from qdash.sdk import QDash

qdash = QDash.from_profile("local")
CHIP_ID = "YOUR_CHIP_ID"
```

## List Executions

```python
executions = qdash.executions.list(chip_id=CHIP_ID, limit=20)
print("Loaded:", len(executions), "Total:", executions.total)
executions.to_dataframe()[["start_at", "name", "status", "execution_id"]]
```

## Find Task Results

API filters search history. Each table shows only the loaded page.

```python
results = qdash.task_results.list(chip_id=CHIP_ID, status="failed", limit=20)
print("Loaded:", len(results), "Total:", results.total)
results.to_dataframe()[["start_at", "task_name", "qid", "status", "task_id"]]
```

## Next Page

`next_page()` preserves the API filters and returns `None` at the end. New records may shift offset-based pages.

```python
next_results = results.next_page()
next_results
```

## Inspect a Result

Copy a task ID from the table to inspect parameters and figures.

```python
task = qdash.task_result("YOUR_TASK_ID")
print(task.message)
display(task.input_parameters.to_dataframe()[["name", "value", "unit", "error"]])
display(task.output_parameters.to_dataframe()[["name", "value", "unit", "error"]])
for figure in task.figures:
    figure.show()
```
