# Read a project

Configure a `local` profile using the [Python reference](../../../docs/reference/clients/python.md). This notebook only reads data.

```python
from qdash.sdk import QDash

qdash = QDash.from_profile("local")
```

```python
for chip in qdash.chips.list():
    print(chip.chip_id, chip.qubit_count)
```

```python
for workflow in qdash.workflows.list():
    print(workflow.name, workflow.chip_id)
```

```python
qdash.close()
```
