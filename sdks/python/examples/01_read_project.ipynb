# Read a project

Run `qdash config init --project-id your-project-id` in a terminal, then fill in `api_token` in the generated file. Use `--base-url` for a remote deployment. If the file already exists, edit it instead. `QDash()` reads its `default_profile`; use `QDash.from_profile("local")` to select a profile explicitly. See the [Python reference](../../../docs/reference/clients/python.md#connect-to-a-deployment) for setup. This notebook only reads data.

```python
from qdash.sdk import QDash

qdash = QDash()
```

```python
for chip in qdash.chips.list():
    print(chip.chip_id, chip.qubit_count)
```

```python
for workflow in qdash.workflows.list():
    print(workflow.name, workflow.chip_id)
```
