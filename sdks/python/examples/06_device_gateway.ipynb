# Export Device Gateway Files

Configure a `local` profile using the [Python reference](../../../docs/reference/clients/python.md). The project needs a master calibration note and chip data. These operations do not start measurements.

```python
from qdash.sdk import QDash

qdash = QDash.from_profile("local")
```

## Create Topology

Choose physical qubits from the chip associated with the latest project calibration note. `device_id` is an output label, not a chip selector.

```python
topology = qdash.device_topology.create(
    name="anemone",
    device_id="anemone",
    qubits=["0", "2", "3"],
)
topology
```

## Save Topology

Existing files are not overwritten. Pass `overwrite=True` to replace them.

```python
topology.save("device_topology.json")
plot = topology.plot()
plot.save("device_topology.png")
plot
```

## Download Qubex Config

The export includes `calib_note.json`. Config and note are separate reads, not a single snapshot. The destination parent must exist.

```python
qdash.qubex.download_config("./qubex-config")
```

## Close Connection

```python
qdash.close()
```
